```python
import requests
from bs4 import BeautifulSoup
url="https://subslikescript.com/movies"
result=requests.get(url)
content=result.text
soup=BeautifulSoup(content,'html.parser')
print(soup.prettify())
```

```python
len(soup.find_all('a',href=True))
```

```python
data=soup.find('article',class_='main-article')
```

```python
len(data.find_all('a',href=True))
```

```python
for i in data.find_all('a'):
     print(i)
```

```python
for i in data.find_all('a'):
     print(i['href'])
```

```python
links=[]
for i in data.find_all('a'):
     links.append("https://subslikescript.com"+i['href'])
```

```python
links
```

```python
for i in links:
    result=requests.get(i)
    content=result.text
    soup=BeautifulSoup(content,'html.parser')
    box=soup.find('article',class_='main-article')
    title=box.find('h1').text.split(":")[0]
    script=box.find('div',class_='full-script').text
    try:
        with open(f'{title}.txt',"w",encoding='utf-8') as file:
            file.write(script)
    except Exception as e:
        print(e)
```
